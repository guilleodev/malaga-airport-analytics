**Flight Routes Analysis: September 2026**

Exploratory data analysis of flight routes, origins, and destinations at Málaga-Costa del Sol Airport using Python, Pandas, and Matplotlib.

### Objectives

- Identify the most frequent flight destinations from Málaga.
- Discover the main airports of origin for flights arriving in Málaga.
- Analyze flight connections between Málaga and other airports.
- Visualize flight activity using real aviation data.

### Datasets

- **Flight List:** OpenSky flight records for September 2026.
- **Airport Codes:** Airport names, countries, and ICAO identifiers.

**Tools:** Python, Pandas, Matplotlib and Jupyter Notebook

## 1. Data Loading

We begin by loading the datasets using Pandas.

The flight dataset contains individual flight records, while the airport dataset provides information about airport locations, names, and country codes.

```python
# ==========================================
# 1. LOAD DATASETS
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

df_airport_codes = pd.read_csv("../data/airport-codes.csv")
df_flight_list_0926 = pd.read_parquet("../data/flight_lists/flight_list_202609.parquet")

print("===== AIRPORTS =====")
display(df_airport_codes.head(5))
print("===== FLIGHTS =====")
display(df_flight_list_0926.head(5))
```

## 2. Dataset Overview

Before analyzing flight routes, we explore the structure of both datasets.

We examine the number of rows and columns, available variables, and data types to understand the information we are working with.

```python
# ==========================================
# 2. DATASET OVERVIEW
# ==========================================

print("===== FLIGHTS =====")
print("Shape:", df_flight_list_0926.shape)
print()
display(df_flight_list_0926.info())

print("\n===== AIRPORTS =====")
print("Shape:", df_airport_codes.shape)
print()
display(df_airport_codes.info())
```

## 3. Data Quality

We check for missing values and duplicated records in both datasets.

This helps us identify potential data quality issues before performing the analysis.

At this stage, we only inspect the data without making modifications.

```python
# ==========================================
# MISSING VALUES
# ==========================================

print("===== FLIGHTS =====")
display(df_flight_list_0926.isna().sum())

print("===== AIRPORTS =====")
display(df_airport_codes.isna().sum())
```

```python
# ==========================================
# DUPLICATES
# ==========================================

print("===== DUPLICATES =====")

print("Flights:", df_flight_list_0926.duplicated().sum())
print("Airports:", df_airport_codes.duplicated().sum())
```

## 4. Malaga Airport Flights

We focus on flights departing from or arriving at Málaga-Costa del Sol Airport.

The ICAO airport code for Málaga is **LEMG**.

We create two DataFrames:

- **Departures:** Flights departing from Málaga.
- **Arrivals:** Flights arriving at Málaga.

This initial analysis uses the `adep` and `ades` columns. Alternative airport fields will be reviewed before drawing final conclusions.

```python
# ==========================================
# 4. MALAGA FLIGHTS
# ==========================================

# Flights departing from Malaga
df_departures_malaga = df_flight_list_0926[df_flight_list_0926["adep"] == "LEMG"].copy()

# Flights arriving at Malaga
df_arrivals_malaga = df_flight_list_0926[df_flight_list_0926["ades"] == "LEMG"].copy()

print("Departures:", len(df_departures_malaga))
print("Arrivals:", len(df_arrivals_malaga))
print("Total Flights SEP26:", len(df_departures_malaga) + len(df_arrivals_malaga))
```

## 5. Destinations: Top 10 Flight Destinations

    Which airports receive the most flights departing from Málaga?

We use `value_counts()` to identify the ten most frequent destinations and visualize the results with Matplotlib.

```python
# ==========================================
# 5. TOP 10 DESTINATIONS
# ==========================================

top_destinations = df_departures_malaga["ades"].value_counts().head(10)

# Merge with airport information
top_destinations = pd.merge(
    top_destinations,
    df_airport_codes[["ident", "name", "municipality", "iso_country"]],
    left_on="ades",
    right_on="ident",
    how="left"
)

top_destinations["municipality"] = (
    top_destinations["municipality"]
    .str.split(",")
    .str[0]
    .str.strip()
)

display(top_destinations)
```

```python
print("===== DATA VISUALIZATION =====")

plt.figure(figsize=(12, 5))

plt.bar(
    top_destinations["municipality"].fillna(top_destinations["name"]),
    top_destinations["count"],
    color="blue"
)

plt.title("Top 10 Destinations from Malaga")
plt.xlabel("Destination City")
plt.ylabel("Number of Flights")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## 6. Origins: Top 10 Flight Origins

    Where do flights arriving in Málaga come from?

We analyze the most frequent departure airports for flights arriving at Málaga and display the results using a bar chart.

```python
# ==========================================
# 6. TOP 10 ORIGINS
# ==========================================

top_origins = df_arrivals_malaga["adep"].value_counts().head(10)

# Merge with airport information
top_origins = pd.merge(
    top_origins,
    df_airport_codes[["ident", "name", "municipality", "iso_country"]],
    left_on="adep",
    right_on="ident",
    how="left"
)

top_origins["municipality"] = (
    top_origins["municipality"]
    .str.split(",")
    .str[0]
    .str.strip()
)

display(top_origins)
```

```python
print("===== DATA VISUALIZATION =====")

plt.figure(figsize=(12, 5))

plt.bar(
    top_origins["municipality"].fillna(top_origins["name"]),
    top_origins["count"],
    color="orange"
)

plt.title("Top 10 Origins to Malaga")
plt.xlabel("Origin City")
plt.ylabel("Number of Flights")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## 7. Flights by Country

In this section, we analyze the countries with the highest number
of recorded flights departing from and arriving at Málaga Airport.

Airport information is combined with flight records using Pandas merge().

### 7.1 Destination: Top 10 Destination Countries

```python
# ==========================================
# 7. FLIGHTS BY COUNTRY
# ==========================================

# Count flights by destination airport
flights_by_destination = (df_departures_malaga["ades"].value_counts().reset_index())

# Merge with airport information
flights_by_destination = pd.merge(
    flights_by_destination,
    df_airport_codes[["ident", "iso_country"]],
    left_on="ades",
    right_on="ident",
    how="left",
    validate="many_to_one"
)

# Group flights by country
top_destination_countries = (
    flights_by_destination.groupby("iso_country")["count"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

display(top_destination_countries)

print("Unmatched airports:", flights_by_destination["iso_country"].isna().sum())
```

```python
print("===== DATA VISUALIZATION =====")

plt.figure(figsize=(10, 5))

plt.bar(
    top_destination_countries.index,
    top_destination_countries.values,
    color="blue"
)

plt.title("Top 10 Destination Countries from Malaga")
plt.xlabel("Country (ISO Code)")
plt.ylabel("Number of Flights")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### 7.2 Origin: Top 10 Origin Countries

```python
# Count flights by origin airport
flights_by_origin = (df_arrivals_malaga["adep"].value_counts().reset_index())

# Merge with airport information
flights_by_origin = pd.merge(
    flights_by_origin,
    df_airport_codes[["ident", "iso_country"]],
    left_on="adep",
    right_on="ident",
    how="left",
    validate="many_to_one"
)

# Group flights by country
top_origin_countries = (
    flights_by_origin.groupby("iso_country")["count"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

display(top_origin_countries)

print("Unmatched airports:", flights_by_origin["iso_country"].isna().sum())
```

```python
print("===== DATA VISUALIZATION =====")

plt.figure(figsize=(10, 5))

plt.bar(
    top_origin_countries.index,
    top_origin_countries.values,
    color="orange"
)

plt.title("Top 10 Origin Countries to Malaga")
plt.xlabel("Country (ISO Code)")
plt.ylabel("Number of Flights")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## 8. Daily Flight Activity

This section explores the daily distribution of flights at Málaga Airport
during September 2026.

We compare the number of recorded arrivals and departures to identify
variations in flight activity throughout the month.

```python
# ==========================================
# 8. DAILY FLIGHT ACTIVITY
# ==========================================

# Convert date column to datetime
df_departures_malaga["dof"] = pd.to_datetime(df_departures_malaga["dof"],errors="coerce")

df_arrivals_malaga["dof"] = pd.to_datetime(df_arrivals_malaga["dof"],errors="coerce")

print("Invalid departure dates:",
      df_departures_malaga["dof"].isna().sum())

print("Invalid arrival dates:",
      df_arrivals_malaga["dof"].isna().sum())
```

```python
# Departures per day
daily_departures = (df_departures_malaga.groupby("dof").size().sort_index())

# Arrivals per day
daily_arrivals = (df_arrivals_malaga.groupby("dof").size().sort_index())

print("===== DAILY DEPARTURES =====")
display(daily_departures.head())

print("===== DAILY ARRIVALS =====")
display(daily_arrivals.head())
```

```python
print("===== DATA VISUALIZATION =====")

plt.figure(figsize=(12, 5))

plt.plot(
    daily_departures.index,
    daily_departures.values,
    label="Departures",
    color="blue"
)

plt.plot(
    daily_arrivals.index,
    daily_arrivals.values,
    label="Arrivals",
    color="orange"
)

plt.title("Daily Flight Activity at Malaga Airport")
plt.xlabel("Date")
plt.ylabel("Number of Flights")

plt.legend()
plt.grid(alpha=0.3)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
print("===== BUSIEST DAYS =====")

print("Day with most departures:")
print(daily_departures.idxmax(), daily_departures.max())

print("\nDay with most arrivals:")
print(daily_arrivals.idxmax(), daily_arrivals.max())

print("\nAverage daily departures:")
print(round(daily_departures.mean(), 2))

print("\nAverage daily arrivals:")
print(round(daily_arrivals.mean(), 2))
```
